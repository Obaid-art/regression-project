## Step 1 · Look at the data
Red rings are **impossible** rows (wrong unit, negative, dead sensor): ticked for you. The purple ring is an **outlier**: possible, but far from the rest. You decide.

```python
#@title Step 1 · choose your group, look, keep or drop  { display-mode: "form" }
out1 = W.Output(); box1 = W.VBox()
def draw_data(ax, ring=True, show_split=False):
    x, y, m = S["x"], S["y"], S["meta"]; keep = S.get("keep", np.array([i + 1 not in {f["row"] for f in S["flags"] if f["kind"] == "impossible"} for i in range(len(x))]))   # before the group decides, the picture is framed without the impossible rows
    lo, hi = np.percentile(y[keep], [0, 100]); pad = 0.08 * (hi - lo + 1e-9)
    if ring and any(y[f["row"] - 1] > hi for f in S["flags"]): hi = hi + pad   # headroom for a ring and its label at the top edge
    if show_split and "xtr" in S: dots(ax, S["xtr"], S["ytr"], label="training rows"); dots(ax, S["xte"], S["yte"], test=True, label="test rows (sealed)"); legend_below(ax)
    else: dots(ax, x[keep], y[keep])
    xs = x[keep]; ax.set(xlim=(xs.min() - 0.05 * (xs.max() - xs.min()), xs.max() + 0.05 * (xs.max() - xs.min())), ylim=(0 if lo >= 0 else lo - pad, hi + pad))
    if ring:
        x0, x1 = ax.get_xlim(); y0, y1 = ax.get_ylim()
        for f in S["flags"]:
            i = f["row"] - 1; inside = x0 <= x[i] <= x1 and y0 <= y[i] <= y1
            px = np.clip(x[i], x0, x1); py = np.clip(y[i], y0, y1); col = RED if f["kind"] == "impossible" else PURPLE
            ax.scatter([px], [py], s=260, facecolor="none", edgecolor=col, linewidth=2.2, zorder=7, marker="o" if inside else "^", clip_on=False)
            place_label(ax, px, py, f"row {f['row']}", col)
    labels(ax)
def choose_group(g):
    m = DATA[str(g)]; x, y = parse(m["csv"]); S.clear(); S.update(group=int(g), meta=m, x=x, y=y, flags=flags(x, y, m), attempts=[], board=[], nets={})
    S["cbs"] = [W.Checkbox(value=(f["kind"] == "impossible"), description=f"drop row {f['row']}: {f['why']}", indent=False, layout=W.Layout(width="98%"), style={"description_width": "0px"}) for f in S["flags"]]
    box1.children = S["cbs"]
    with out1:
        clear_output(wait=True); print(f"Group {g}: {m['name']}\n{m['story']}\n{len(x)} rows: {m['x']} ({m['xu']}) against {m['y']} ({m['yu']}).\n")
        fig, ax = plt.subplots(figsize=(11.5, 6)); draw_data(ax); ax.set_title(f"group {g}: {m['name']} (red = impossible, purple = look)"); finish(fig)
        for f in S["flags"]: print(f"  row {f['row']:>2}  {m['x']} = {g3(x[f['row'] - 1])}, {m['y']} = {g3(y[f['row'] - 1])}   ->  {f['why']}")
def keep_choice(_=None):
    S["drop"] = {f["row"] for f, cb in zip(S["flags"], S["cbs"]) if cb.value}; S["keep"] = np.array([i + 1 not in S["drop"] for i in range(len(S["x"]))])
    S["drop_notes"] = {f["row"]: f["why"] for f in S["flags"] if f["row"] in S["drop"]}
    with out1:
        clear_output(wait=True); fig, ax = plt.subplots(figsize=(11.5, 6)); draw_data(ax, ring=False); ax.set_title(f"group {S['group']}: {int(S['keep'].sum())} rows kept"); finish(fig)
        print(f"You kept {int(S['keep'].sum())} rows and dropped {len(S['drop'])}: {sorted(S['drop']) if S['drop'] else 'none'}.")
        kept_looks = [f for f in S["flags"] if f["kind"] == "look" and f["row"] not in S["drop"]]
        if kept_looks: print("You kept the outlier in row", ", ".join(str(f["row"]) for f in kept_looks) + ". Fine, if you can say why.")
S.setdefault("intuition", {})
i1 = W.Textarea(description="", placeholder="No right or wrong answer. From the picture alone: would a straight line describe this? What shape do you expect, and why?", layout=W.Layout(width="95%", height="80px"))
i1.observe(lambda ch: S.setdefault("intuition", {}).__setitem__("shape", ch["new"]), names="value")
dd = W.Dropdown(options=[(f"Group {g}: {m['name']}", int(g)) for g, m in DATA.items()], description="your group", style={"description_width": "90px"}, layout=W.Layout(width="60%"))
dd.observe(lambda ch: choose_group(ch["new"]), names="value"); btn1 = W.Button(description="Keep my choice", button_style="primary"); btn1.on_click(keep_choice)
display(dd, out1, HTML("<b>Your intuition, before any fitting</b> (no right or wrong answer): would a straight line describe this picture? What shape do you expect, and why?"), i1, box1, btn1); choose_group(dd.value)
```

## Step 1b · What the outlier does
Press the button: the line is fitted twice on the same training rows, with the purple row and without it, and judged on the same sealed rows.

```python
#@title Step 1b · what the outlier does  { display-mode: "form" }
out1b = W.Output()
def line_on(xtr, ytr):
    mx, sx, my, sy = xtr.mean(), xtr.std(), ytr.mean(), ytr.std(); xp, yp = (xtr - mx) / sx, (ytr - my) / sy; mstd = float((xp * yp).sum() / (xp * xp).sum())
    f = lambda v, mstd=mstd, mx=mx, sx=sx, my=my, sy=sy: ((np.asarray(v, float) - mx) / sx * mstd) * sy + my; b = mstd * sy / sx; return f, S["my"] if False else (my - b * mx), b
def outlier_effect(_=None):
    looks = [f for f in S["flags"] if f["kind"] == "look"]
    with out1b:
        clear_output(wait=True)
        if not looks: print("No outlier was flagged in this dataset."); return
    row = looks[0]["row"]; base = np.array([i + 1 not in {f["row"] for f in S["flags"] if f["kind"] == "impossible"} for i in range(len(S["x"]))])
    x, y, idx = S["x"][base], S["y"][base], np.arange(len(S["x"]))[base]; test = np.arange(len(x)) % 4 == 3; m = S["meta"]; res = {}
    for label, drop in (("with the outlier", False), ("without it", True)):
        tr = ~test & ~(np.full(len(x), drop) & (idx == row - 1)); f, a, b = line_on(x[tr], y[tr])
        res[label] = dict(train=r2(y[tr], f(x[tr])), test=r2(y[test], f(x[test])), a=a, b=b, xtr=x[tr], ytr=y[tr], eq=f"{m['y']} = {g3(a)} {'−' if b < 0 else '+'} {g3(abs(b))} · {m['x']}", f=f)
    S["outlier_effect"] = {k: {kk: vv for kk, vv in v.items() if kk not in ("f", "xtr", "ytr")} for k, v in res.items()}; S["outlier_effect"]["row"] = row
    grid = np.linspace(x.min(), x.max(), 200)
    def draw_outlier_effect(fig):
        a1, a2 = fig.subplots(1, 2, sharey=True)
        for ax, (label, r) in zip((a1, a2), res.items()):
            dots(ax, r["xtr"], r["ytr"], label="training rows"); dots(ax, x[test], y[test], test=True, label="sealed test rows"); ax.plot(grid, r["f"](grid), color=INK, lw=3, label="the line")
            if label.startswith("with"): ax.scatter([S["x"][row - 1]], [S["y"][row - 1]], s=300, facecolor="none", edgecolor=PURPLE, linewidth=2.4, zorder=7)
            labels(ax); legend_below(ax, ncol=2); ax.set_title(f"{label}:  test R² = {r['test']:.3f}")
    S["draw_outlier_effect"] = draw_outlier_effect
    with out1b:
        fig = plt.figure(figsize=(14.5, 6)); draw_outlier_effect(fig); finish(fig)
        w, wo = res["with the outlier"], res["without it"]
        print(f"with the outlier:   {w['eq']}      test R² = {w['test']:.3f}")
        print(f"without it:         {wo['eq']}      test R² = {wo['test']:.3f}")
        print(f"One row changed the slope by {abs(w['b'] - wo['b']) / max(abs(wo['b']), 1e-9) * 100:.0f} % and the sealed score by {wo['test'] - w['test']:+.3f}. Now decide, and say why.")
i2 = W.Textarea(description="", placeholder="Keep the purple row, or drop it? Why? (Your call; what matters is the reason.)", layout=W.Layout(width="95%", height="70px"))
i2.observe(lambda ch: S.setdefault("intuition", {}).__setitem__("outlier", ch["new"]), names="value")
btn1b = W.Button(description="Show what the outlier does", button_style="primary"); btn1b.on_click(outlier_effect); display(btn1b, out1b, HTML("<b>Your decision</b>: keep it or drop it, and why?"), i2)
```

## Step 2 · Split and standardise
Every fourth kept row is **sealed** as the test set. Then the bank teller: x′ = (x − mean) / std, with the **training** rows' mean and std.

```python
#@title Step 2 · split and standardise  { display-mode: "form" }
out2 = W.Output()
def split(_=None):
    if "keep" not in S: keep_choice()
    x, y = S["x"][S["keep"]], S["y"][S["keep"]]; test = np.arange(len(x)) % 4 == 3
    S.update(xtr=x[~test], ytr=y[~test], xte=x[test], yte=y[test]); S.update(mx=S["xtr"].mean(), sx=S["xtr"].std(), my=S["ytr"].mean(), sy=S["ytr"].std())
    S["zx"] = lambda v: (v - S["mx"]) / S["sx"]; S["zy"] = lambda v: (v - S["my"]) / S["sy"]; S["unzy"] = lambda v: v * S["sy"] + S["my"]
    S["grid"] = np.linspace(x.min(), x.max(), 200); S["board"] = []; S["nets"] = {}; S.pop("line", None); S.pop("compare", None)
    m = S["meta"]
    with out2:
        clear_output(wait=True); fig, (a1, a2) = plt.subplots(1, 2, figsize=(14.5, 5.6), gridspec_kw={"width_ratios": [1.15, 1]})
        draw_data(a1, ring=False, show_split=True); a1.set_title(f"{len(S['xtr'])} training rows, {len(S['xte'])} sealed test rows")
        a2.scatter(S["zx"](S["xtr"]), np.ones(len(S["xtr"])), s=60, color=HALO, edgecolor=BLUE); a2.scatter(S["zy"](S["ytr"]), np.zeros(len(S["ytr"])), s=60, color=HALO, edgecolor=BLUE)
        a2.set(yticks=[1, 0], yticklabels=[f"{m['x']}′", f"{m['y']}′"], ylim=(-0.6, 1.6), xlabel="standardised value: (value − mean) / std"); a2.set_title("one currency: both live between about −2 and +2"); a2.axvline(0, color=GREY, lw=1)
        finish(fig)
        x0, y0 = S["xtr"][0], S["ytr"][0]
        print(f"{m['x']}: mean = {g3(S['mx'])} {m['xu']}, std = {g3(S['sx'])} {m['xu']}      {m['y']}: mean = {g3(S['my'])} {m['yu']}, std = {g3(S['sy'])} {m['yu']}")
        print(f"first training row, by substitution:   x′ = ({g3(x0)} − {g3(S['mx'])}) / {g3(S['sx'])} = {S['zx'](x0):+.2f}      y′ = ({g3(y0)} − {g3(S['my'])}) / {g3(S['sy'])} = {S['zy'](y0):+.2f}")
btn2 = W.Button(description="Split and standardise", button_style="primary"); btn2.on_click(split); display(btn2, out2)
```

## Step 2b · Why standardising is a must
The same network, the same gradient descent, the same 300 epochs, twice: on the raw numbers and on the standardised ones.

```python
#@title Step 2b · why standardising is a must  { display-mode: "form" }
out2b = W.Output()
def std_effect(_=None):
    if "xtr" not in S: split()
    res = {}
    for label, raw in (("raw numbers", True), ("standardised", False)):
        net = Net([8, 8], seed=7); X = (S["xtr"] if raw else S["zx"](S["xtr"]))[:, None]; y = S["ytr"] if raw else S["zy"](S["ytr"])
        with np.errstate(all="ignore"):
            for e in range(300): net.epoch(X, y, lr=0.05, plain=True)
        f = (lambda v, net=net: net.predict(np.asarray(v, float)[:, None])) if raw else (lambda v, net=net: S["unzy"](net.predict(S["zx"](np.asarray(v, float))[:, None])))
        tr, te = r2(S["ytr"], f(S["xtr"])), r2(S["yte"], f(S["xte"])); res[label] = dict(f=f, train=float(np.nan_to_num(tr, nan=-9.0)), test=float(np.nan_to_num(te, nan=-9.0)))
    S["std_effect"] = {k: {kk: vv for kk, vv in v.items() if kk != "f"} for k, v in res.items()}
    def draw_std_effect(fig):
        a1, a2 = fig.subplots(1, 2, sharey=True)
        for ax, (label, r) in zip((a1, a2), res.items()):
            dots(ax, S["xtr"], S["ytr"], label="training rows"); dots(ax, S["xte"], S["yte"], test=True, label="sealed test rows"); ax.plot(S["grid"], np.clip(r["f"](S["grid"]), *np.percentile(S["y"][S["keep"]], [0, 100]) * np.array([0.0, 1.0]) + np.array([-0.2, 0.3]) * np.ptp(S["y"][S["keep"]])), color=GREEN, lw=3, label="the network")
            labels(ax); legend_below(ax, ncol=2); ax.set_title(f"{label}:  test R² = {max(r['test'], -9):.2f}")
    S["draw_std_effect"] = draw_std_effect
    with out2b:
        clear_output(wait=True); fig = plt.figure(figsize=(14.5, 6)); draw_std_effect(fig); finish(fig)
        print(f"raw numbers:    test R² = {res['raw numbers']['test']:.2f}       standardised:    test R² = {res['standardised']['test']:.2f}")
        print("Same network, same recipe (gradient descent), same epochs. On raw numbers every step is the wrong size: too big for one weight, too small for another, and the bends switch off. One currency first.")
btn2b = W.Button(description="Train twice: raw against standardised", button_style="primary"); btn2b.on_click(std_effect); display(btn2b, out2b)
```

## Step 3 · The line
One knob: on standardised data the line passes through the middle, so only the slope m is free. Move the slider, then let calculus find the bottom of the bowl.

```python
#@title Step 3 · the line  { display-mode: "form" }
out3 = W.Output()
def line_fn(mstd): return lambda v: S["unzy"](mstd * S["zx"](v))
def draw_line(ax, f, which="train", sticks=True, label="the line"):
    xs, ys = (S["xtr"], S["ytr"]) if which == "train" else (S["xte"], S["yte"])
    dots(ax, xs, ys, test=(which == "test")); ax.plot(S["grid"], f(S["grid"]), color=INK, lw=3, label=label)
    if sticks:
        for a, b in zip(xs, ys): ax.plot([a, a], [b, f(a)], color=RED, lw=2, alpha=0.8)
    labels(ax); ax.set_title(f"{'training rows' if which == 'train' else 'the sealed test rows'}: R² = {r2(ys, f(xs)):.3f}")
def show_line(mstd):
    if "xtr" not in S: split()
    xp, yp = S["zx"](S["xtr"]), S["zy"](S["ytr"]); L = float(((yp - mstd * xp) ** 2).sum()); S["attempts"].append((mstd, L)); f = line_fn(mstd)
    with out3:
        clear_output(wait=True); fig, (a1, a2) = plt.subplots(1, 2, figsize=(14.5, 5.6), gridspec_kw={"width_ratios": [1.15, 1]})
        draw_line(a1, f); a1.set_title(f"attempt {len(S['attempts'])}:  m = {mstd:g},  loss L = {L:.2f}")
        mm = np.linspace(-2, 2, 200); a2.plot(mm, [float(((yp - v * xp) ** 2).sum()) for v in mm], color=BLUE, lw=2.5, alpha=0.35, label="the whole bowl")
        a2.scatter([a for a, _ in S["attempts"]], [b for _, b in S["attempts"]], s=70, color=ADDED, edgecolor=RED, linewidth=1.5, zorder=5, label="your attempts"); a2.scatter([mstd], [L], s=160, color=RED, zorder=6)
        a2.set(xlabel="slope m (standardised)", ylabel="loss L"); a2.set_title("the bowl: loss against the slope"); a2.legend(loc="upper center"); finish(fig)
        print(f"training R² = {r2(S['ytr'], f(S['xtr'])):.3f}     test R² = {r2(S['yte'], f(S['xte'])):.3f}")
def fit_line(_=None):
    if "xtr" not in S: split()
    xp, yp = S["zx"](S["xtr"]), S["zy"](S["ytr"]); mstd = float((xp * yp).sum() / (xp * xp).sum()); f = line_fn(mstd); m = S["meta"]
    b = mstd * S["sy"] / S["sx"]; a = S["my"] - b * S["mx"]
    S["line"] = dict(mstd=mstd, a=a, b=b, train=r2(S["ytr"], f(S["xtr"])), test=r2(S["yte"], f(S["xte"])), loss=float(((yp - mstd * xp) ** 2).sum()), eq=f"{m['y']} = {g3(a)} {'−' if b < 0 else '+'} {g3(abs(b))} · {m['x']}", f=f)
    with out3:
        clear_output(wait=True); fig, (a1, a2) = plt.subplots(1, 2, figsize=(14.5, 5.6), sharey=True)
        draw_line(a1, f); draw_line(a2, f, which="test"); finish(fig)
        print(f"m = Σx′y′ / Σx′² = {g3((xp * yp).sum())} / {g3((xp * xp).sum())} = {mstd:.3f}   (standardised slope; loss at the bottom = {S['line']['loss']:.2f})")
        print(f"back in real units:   {S['line']['eq']}")
        print(f"training R² = {S['line']['train']:.3f}     test R² = {S['line']['test']:.3f}")
sl = W.FloatSlider(value=0.5, min=-2, max=2, step=0.05, description="slope m", continuous_update=False, layout=W.Layout(width="60%"), style={"description_width": "70px"})
sl.observe(lambda ch: show_line(ch["new"]), names="value"); btn3 = W.Button(description="Let calculus find the bottom", button_style="primary"); btn3.on_click(fit_line); display(sl, btn3, out3)
with out3: print("move the slider to make your first attempt")
```

## Step 4 · The network board: play
Five knobs. Train as many times as you like (each run takes seconds). The board keeps every run. Then **pick one**, and say why.

Two challenges: try to make the network **memorise** (training score far above the test score; not every dataset lets you), and find the **smallest** network that keeps your best test score.

```python
#@title Step 4 · the network board  { display-mode: "form" }
out4 = W.Output(); outb = W.Output()
def best_run(): return max(S["board"], key=lambda r: r["test"]) if S["board"] else None
def chosen_run():
    """the run the group picked on the board; the best test score if they have not picked yet"""
    if S.get("pick_run") in S.get("nets", {}): return S["nets"][S["pick_run"]]
    return best_run()
def badges():
    out = []
    if any(r["train"] - r["test"] > 0.10 for r in S["board"]): out.append("challenge 1 done: you made it memorise")
    b = best_run()
    if b and any(r is not b and r["weights"] < b["weights"] and r["test"] >= b["test"] - 0.02 for r in S["board"]): out.append("challenge 2 done: a smaller network keeps the best score")
    return out
def show_board():
    with outb:
        clear_output(wait=True)
        if not S["board"]: print("the board is empty: train something"); return
        best = best_run(); pick = S.get("pick_run")
        def tag(r):
            t = []
            if r is best: t.append("★ best test")
            if r["run"] == pick: t.append("your pick")
            if r["train"] - r["test"] > 0.10: t.append("memorising")
            elif r["train"] < 0.5: t.append("too simple")
            return ", ".join(t)
        rows = "".join(f"<tr style='background:{'#E3F3EC' if r is best else ('#FFF4E0' if r['run'] == pick else 'white')}'><td>{r['run']}</td><td>{r['layers']}</td><td>{r['neurons']}</td><td>{r['epochs']}</td><td>{r['lr']:g}</td><td>{r['batch']}</td><td>{r['weights']}</td><td style='color:{BLUE};font-weight:bold'>{r['train']:.3f}</td><td style='color:{AMBER};font-weight:bold'>{r['test']:.3f}</td><td>{tag(r)}</td></tr>" for r in S["board"])
        done = badges()
        display(HTML(f"<table style='font-size:15px;border-collapse:collapse'><tr><th>run</th><th>layers</th><th>neurons</th><th>epochs</th><th>learning rate</th><th>batch</th><th>weights</th><th>training R²</th><th>test R²</th><th></th></tr>{rows}</table>"
                     + "".join(f"<p style='color:{GREEN};font-weight:bold'>{d}</p>" for d in done)))
        pick_dd.options = [(f"run {r['run']}: {r['layers']} × {r['neurons']}, {r['epochs']} epochs, test R² {r['test']:.3f}", r["run"]) for r in S["board"]]
        if pick in S["nets"]: pick_dd.value = pick
def draw_net(ax, run, which="both"):
    r = S["nets"][run]
    if which in ("both", "train"): dots(ax, S["xtr"], S["ytr"], label="training rows")
    if which in ("both", "test"): dots(ax, S["xte"], S["yte"], test=True, label="test rows (sealed)")
    ax.plot(S["grid"], r["yg"], color=GREEN, lw=3.2, label=f"run {run}: {r['layers']} × {r['neurons']}, {r['epochs']} epochs"); labels(ax); legend_below(ax)
def train_cfg(L, n, E, lr=0.01, batch=16, live=True):
    if "xtr" not in S: split()
    net = Net([n] * L, seed=len(S["board"]) + 1); X = S["zx"](S["xtr"])[:, None]; y = S["zy"](S["ytr"]); Xg = S["zx"](S["grid"])[:, None]
    with np.errstate(all="ignore"):
        for e in range(1, E + 1):
            net.epoch(X, y, lr=lr, batch=batch)
            if live and (e % max(1, E // 10) == 0 or e == E):
                with out4:
                    clear_output(wait=True); fig, ax = plt.subplots(figsize=(11.5, 6)); dots(ax, S["xtr"], S["ytr"], label="training rows"); dots(ax, S["xte"], S["yte"], test=True, label="test rows (sealed)"); ax.plot(S["grid"], S["unzy"](net.predict(Xg)), color=GREEN, lw=3.2, label="the network")
                    labels(ax); legend_below(ax); ax.set_title(f"{L} layer(s) × {n} neurons · learning rate {lr:g} · batch {batch} · epoch {e} of {E}"); finish(fig)
    f = lambda v: S["unzy"](net.predict(S["zx"](np.asarray(v, float))[:, None])); run = len(S["board"]) + 1
    tr, te = r2(S["ytr"], f(S["xtr"])), r2(S["yte"], f(S["xte"]))
    row = dict(run=run, layers=L, neurons=n, epochs=E, lr=lr, batch=batch, weights=net.n, train=float(np.nan_to_num(tr, nan=-9.0)), test=float(np.nan_to_num(te, nan=-9.0)))
    S["board"].append(row); S["nets"][run] = dict(row, yg=np.nan_to_num(f(S["grid"]), nan=0.0), f=f); show_board(); return row
layers = W.IntSlider(value=1, min=1, max=3, description="layers"); neurons = W.SelectionSlider(options=[1, 2, 4, 8, 16, 32, 64], value=4, description="neurons"); epochs = W.SelectionSlider(options=[50, 100, 300, 1000, 3000, 10000], value=300, description="epochs")
lrate = W.SelectionSlider(options=[0.001, 0.003, 0.01, 0.03, 0.1], value=0.01, description="learning rate", style={"description_width": "100px"}); bsize = W.SelectionSlider(options=[4, 8, 16, 32], value=16, description="batch")
btn4 = W.Button(description="▶ Train", button_style="primary"); btn4.on_click(lambda _: train_cfg(int(layers.value), int(neurons.value), int(epochs.value), float(lrate.value), int(bsize.value)))
pick_dd = W.Dropdown(options=[], description="my pick", style={"description_width": "80px"}, layout=W.Layout(width="70%"))
def set_pick(ch):
    if ch["new"] is not None: S["pick_run"] = int(ch["new"]); show_board()
pick_dd.observe(set_pick, names="value")
i5 = W.Textarea(description="", placeholder="Why this run and not another? (The reason is what counts: for example, the highest test score, or the same score with fewer weights.)", layout=W.Layout(width="95%", height="70px"))
i5.observe(lambda ch: S.setdefault("intuition", {}).__setitem__("pick", ch["new"]), names="value")
display(W.HBox([layers, neurons, epochs]), W.HBox([lrate, bsize]), btn4, out4, outb, HTML("<b>Your pick</b>: which run goes forward to the comparison, and why?"), pick_dd, i5); show_board()
```

## Step 5 · Three candidates on the sealed rows
The line, your best network, and the physical law where one exists, judged on the same sealed rows. First write down your guess.

```python
#@title Step 5 · three candidates on the sealed rows  { display-mode: "form" }
LAWS = {"lin": (lambda x: np.c_[np.ones_like(x), x], "none"), "sqrt": (lambda x: np.c_[np.sqrt(np.maximum(x, 0))], "none"), "exp": (lambda x: np.c_[np.ones_like(x), x], "log"),
        "arrhenius": (lambda x: np.c_[np.ones_like(x), 1 / np.maximum(x, 1e-9)], "log"), "torricelli": (lambda x: np.c_[np.ones_like(x), x], "sqrt"), "quad": (lambda x: np.c_[np.ones_like(x), x ** 2], "none"), "ergun": (lambda x: np.c_[x, x ** 2], "none")}
def fit_law():
    kind = S["meta"]["law"]
    if kind == "none": return None
    design, tr = LAWS[kind]; x, y = S["xtr"], S["ytr"]
    if tr == "log": ok = y > 0; coef = np.linalg.lstsq(design(x[ok]), np.log(y[ok]), rcond=None)[0]; f = lambda v: np.exp(design(np.asarray(v, float)) @ coef)
    elif tr == "sqrt": ok = y >= 0; coef = np.linalg.lstsq(design(x[ok]), np.sqrt(y[ok]), rcond=None)[0]; f = lambda v: np.maximum(design(np.asarray(v, float)) @ coef, 0) ** 2
    else: coef = np.linalg.lstsq(design(x), y, rcond=None)[0]; f = lambda v: design(np.asarray(v, float)) @ coef
    return dict(kind=kind, text=S["meta"]["law_text"], coef=[float(c) for c in coef], n=len(coef), f=f, train=r2(S["ytr"], f(S["xtr"])), test=r2(S["yte"], f(S["xte"])))
out5 = W.Output()
def draw_compare(ax):
    c = S["compare"]; dots(ax, S["xtr"], S["ytr"], label="training rows"); dots(ax, S["xte"], S["yte"], test=True, label="test rows (sealed)")
    for r, f, col in zip(c["rows"], c["fs"], c["colors"]): ax.plot(S["grid"], f(S["grid"]), color=col, lw=3, label=f"{r['name'].split(' (')[0]}: test R² {r['test']:.3f}")
    labels(ax); legend_below(ax, ncol=2); ax.set_title("three candidates, judged on the sealed rows")
def compare(_=None):
    if "line" not in S: fit_line()
    if not S["board"]:
        with out5: print("no network yet: training a default one (2 layers × 8 neurons, 1000 epochs) for you")
        train_cfg(2, 8, 1000, live=False)
    best = chosen_run(); law = fit_law(); linear_law = law is not None and law["kind"] == "lin"
    rows = [dict(name="the line" + (f" (which is also the physics here: {law['text']})" if linear_law else ""), params=2, train=S["line"]["train"], test=S["line"]["test"], f=S["line"]["f"], color=INK),
                                           dict(name=f"the network (your pick, run {best['run']}: {best['layers']} × {best['neurons']}, {best['epochs']} epochs)", params=best["weights"], train=best["train"], test=best["test"], f=S["nets"][best["run"]]["f"], color=GREEN)]
    bb = best_run(); note = "" if bb is best else f" You picked run {best['run']} (test R² {best['test']:.3f}); the board's best test score was run {bb['run']} ({bb['test']:.3f})."
    if law and not linear_law: rows.append(dict(name=f"the physical law ({law['text']})", params=law["n"], train=law["train"], test=law["test"], f=law["f"], color=PURPLE))
    tol = 0.02   # on fifteen test rows, two hundredths of R² is noise
    if law and not linear_law and law["test"] >= best["test"] - tol and law["test"] >= S["line"]["test"] - tol: verdict = f"The physical law wins on the sealed rows: with {law['n']} number(s) it matches or beats the network with {best['weights']}. Same score, far fewer numbers, and every number has a meaning."
    elif S["line"]["test"] >= best["test"] - tol: verdict = "The line wins on the sealed rows: with 2 numbers it matches the network with " + str(best["weights"]) + (f". The physics ({law['text']}) says a line, too" if linear_law else "") + ". Same score, far fewer numbers."
    else: verdict = f"The network wins on the sealed rows, by {best['test'] - S['line']['test']:.2f} in R² over the line: the shape is not a line, and no law was offered. It earns its {best['weights']} weights here."
    verdict = verdict + note
    S["compare"] = dict(rows=[{k: v for k, v in r.items() if k != "f"} for r in rows], verdict=verdict, law=({k: v for k, v in law.items() if k != "f"} if law else None), fs=[r["f"] for r in rows], colors=[r["color"] for r in rows])
    with out5:
        clear_output(wait=True); fig, ax = plt.subplots(figsize=(11.5, 6)); draw_compare(ax); finish(fig)
        display(HTML("<table style='font-size:14px;border-collapse:collapse'><tr><th>model</th><th>numbers it needs</th><th>training R²</th><th>test R²</th></tr>" + "".join(f"<tr><td>{r['name']}</td><td>{r['params']}</td><td>{r['train']:.3f}</td><td style='font-weight:bold'>{r['test']:.3f}</td></tr>" for r in rows) + f"</table><p style='font-size:16px;color:{GREEN}'><b>Verdict.</b> {verdict}</p>"))
i3 = W.Textarea(description="", placeholder="No right or wrong answer. Before pressing: which will score best on the sealed rows, the line, the network, or a physical law? Why?", layout=W.Layout(width="95%", height="70px"))
i3.observe(lambda ch: S.setdefault("intuition", {}).__setitem__("guess", ch["new"]), names="value")
i4 = W.Textarea(description="", placeholder="No right or wrong answer. Two models score the same; one needs 2 numbers, the other 97. Which would you hand to the plant engineer, and why?", layout=W.Layout(width="95%", height="70px"))
i4.observe(lambda ch: S.setdefault("intuition", {}).__setitem__("hand", ch["new"]), names="value")
btn5 = W.Button(description="Compare the three", button_style="primary"); btn5.on_click(compare)
display(HTML("<b>Your guess, before pressing</b> (no right or wrong answer): which will score best on the sealed rows, and why?"), i3, btn5, out5, HTML("<b>After seeing it</b> (no right or wrong answer): same score, 2 numbers against 97. Which would you hand to the plant engineer, and why?"), i4)
```
